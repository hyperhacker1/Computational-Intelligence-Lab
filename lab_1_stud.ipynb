# Example 1

Print the prime numbers from 1 to 100.

```python
# Example 1
for i in range(101):
  prime=True
  j=2
  while prime and j<i:
    if (i % (j))==0:
      prime=False
    j+=1
  if prime:
    print(i)
```

# Task 1

Modify previous example and define function for testing if the number is a prime number.

```python
def is_prime(num):
  if num < 2:
    return False
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      return False
  return True

print('Prime numbers from 1 to 100:')
for i in range(1, 101):
  if is_prime(i):
    print(i)
```

# Task 2

Modify the previous task and define class finding the prime numbers in selected range.


```python
class PrimeFinder:
    def __init__(self):
        pass

    def is_prime(self, num):
        if num < 2:
            return False
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                return False
        return True

    def find_primes_in_range(self, start, end):
        primes = []
        for num in range(start, end + 1):
            if self.is_prime(num):
                primes.append(num)
        return primes

# Example usage for Task 2
prime_finder = PrimeFinder()
start_range = 1
end_range = 200
found_primes = prime_finder.find_primes_in_range(start_range, end_range)
print(f'Prime numbers between {start_range} and {end_range}:')
print(found_primes)
```

## Task 5

Rencode labels to three numerical colums according to the table:

Label | Output[0] | Output[1] | Output[2]
:--- | :---: | :---: | :---: |
Iris-versicolor | 1 | 0 | 0
Iris-virginica  | 0 | 1 | 0
Iris-setosa     | 0 | 0 | 1

so that the first row of the Iris dataset will look like ``5.1 3.5 1.4 0.2 0 0 1``  



```python
labels = iris_dataset[:, -1]

features = iris_dataset[:, :-1].astype(float)

label_mapping = {
    'Iris-versicolor': [1, 0, 0],
    'Iris-virginica': [0, 1, 0],
    'Iris-setosa': [0, 0, 1]
}

encoded_labels = np.array([label_mapping[label] for label in labels])

iris_dataset_enc = np.concatenate((features, encoded_labels), axis=1)

print(iris_dataset_enc)
```

## Task 6

Split the ``iris_dataset_enc`` into two numpy arrays. The ``iris_features`` will contain four numerical columns containing features and the ``iris_desired`` will contain three numerical outputs containing desired outputs.

```python
iris_features = iris_dataset_enc[:, :4]
iris_desired = iris_dataset_enc[:, 4:]

print('iris_features:')
print(iris_features)
print('\niris_desired:')
print(iris_desired)
```

## Task 7

Normalize features in the ``iris_dataset`` into the range of $<0,1>$ using min-max norm according to the following formula:

$$ y = \frac{y_{max} - y_{min}}{x_{max}-x_{min}}(x - x_{min}) + y_{min} $$

where $x$ is raw value and $y$ is the normalised value. The $y_{min}$ and $y_{max}$ determine the output range. The $x_{max}$ and $x_{min}$ denote the input range.

For this task $y_{max}=1$, $y_{min}=0$. The $x_{max}$ and $x_{min}$ must be obtained from the dataset, for each column separately.

The result prints with limited precision to two decimal digits (0.00).

```python
x_min = np.min(iris_features, axis=0)
x_max = np.max(iris_features, axis=0)
y_min = 0
y_max = 1

print(f"x_min: {x_min}")
print(f"x_max: {x_max}")
print(f"y_min: {[y_min]*4}")
print(f"y_max: {[y_max]*4}")
```

```python
iris_features_norm = ((y_max - y_min) / (x_max - x_min)) * (iris_features - x_min) + y_min

np.set_printoptions(precision=2, suppress=True)
for row in iris_features_norm:
    print(' '.join([f'{val:.2f}' for val in row]))
```

# Task 8

Create a scatterplot by using the first and the second column of data, so that each row composes point colored by a label.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert iris_desired (one-hot encoded) back to single labels for coloring
def get_label(one_hot_vector):
    if np.array_equal(one_hot_vector, [1., 0., 0.]): return 'Iris-versicolor'
    if np.array_equal(one_hot_vector, [0., 1., 0.]): return 'Iris-virginica'
    if np.array_equal(one_hot_vector, [0., 0., 1.]): return 'Iris-setosa'

# Apply the conversion to create a list of labels
iris_labels = [get_label(row) for row in iris_desired]

# Create a DataFrame for easier plotting with seaborn
import pandas as pd
df_plot = pd.DataFrame({
    'feature_1': iris_features_norm[:, 0],
    'feature_2': iris_features_norm[:, 1],
    'species': iris_labels
})

plt.figure(figsize=(10, 7))
sns.scatterplot(x='feature_1', y='feature_2', hue='species', data=df_plot, s=100, alpha=0.8)
plt.title('Iris Dataset: Feature 1 vs Feature 2 (Normalized)')
plt.xlabel('Normalized Feature 1')
plt.ylabel('Normalized Feature 2')
plt.grid(True)
plt.legend(title='Species')
plt.show()
```

# Task 9

The sigmoid function is frequently used activation of a neuron. It is defined by

$$y=\sigma(x)=\frac{1}{1+e^{-x}}$$

Draw a graf of this function for $x\in <-8,8>$.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-8, 8, 100)  # Generate 100 points between -8 and 8
y = sigmoid(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y)
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel('y = \u03c3(x)')
plt.grid(True)
plt.show()
```

# Task 10

The RELU function is frequently used activation function of neurons in deep neural networks. It is defined by

$$y=\mbox{relu}(x)=\mbox{max}(0, x)$$

Draw a graf of this function for $x\in <-10,10>$.

```python
import numpy as np
import matplotlib.pyplot as plt

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 100)  # Generate 100 points between -10 and 10
y = relu(x)

plt.figure(figsize=(8, 6))
plt.plot(x, y)
plt.title('ReLU Function')
plt.xlabel('x')
plt.ylabel('y = max(0, x)')
plt.grid(True)
plt.show()
```

# Task 11

The Gaussian function is frequently used as an activation function of RBF neurons. It is defined by the equation

$$y=g(x)=e^{\frac{-x^2}{2\sigma^2}}$$

Draw a graf of this function for $x\in <-10,10>$ and $\sigma=\{0.2,0.5,1,2,5\}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def gaussian(x, sigma):
    return np.exp(-x**2 / (2 * sigma**2))

x = np.linspace(-10, 10, 100) # Generate 100 points between -10 and 10
sigmas = [0.2, 0.5, 1, 2, 5]

plt.figure(figsize=(10, 7))

for sigma in sigmas:
    y = gaussian(x, sigma)
    plt.plot(x, y, label=fr'$\sigma = {sigma}$', linestyle='-')

plt.title(r'Gaussian Function for Different $\sigma$ Values')
plt.xlabel('x')
plt.ylabel('y = g(x)')
plt.grid(True)
plt.legend()
plt.show()
```

# Task 12

Draw response of the two input perceptron neuron defined by

$$y=\sigma(\sum_{i=0}^{1}w_ix_i + b)$$

where $w_0=10$,$w_1=10$, $b=-10$.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Perceptron parameters
w0 = 10
w1 = 10
b = -10

# Create a meshgrid for x0 and x1 inputs
x0 = np.linspace(-1, 2, 50)
x1 = np.linspace(-1, 2, 50)
X0, X1 = np.meshgrid(x0, x1)

# Calculate the weighted sum and apply the sigmoid function
Z = sigmoid(w0 * X0 + w1 * X1 + b)

# Plot the 3D surface
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X0, X1, Z, cmap='viridis')

ax.set_title('Two-Input Perceptron Response')
ax.set_xlabel('Input x0')
ax.set_ylabel('Input x1')
ax.set_zlabel('Output y')

plt.show()
```

# Task 13

Draw response of the two input RBF neuron defined by

$$y=g(\sqrt{\sum_{i=0}^{1}(x_i-c_i)^2})$$

where $c_0=0.5$, $c_1=0.5$, $\sigma=0.15$.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the Gaussian (RBF) function
def gaussian(r, sigma):
    return np.exp(-r**2 / (2 * sigma**2))

# RBF neuron parameters
c0 = 0.5
c1 = 0.5
sigma = 0.15

# Create a meshgrid for x0 and x1 inputs
x0 = np.linspace(-2, 2, 50)
x1 = np.linspace(-2, 2, 50)
X0, X1 = np.meshgrid(x0, x1)

# Calculate the Euclidean distance from the center (c0, c1)
r = np.sqrt((X0 - c0)**2 + (X1 - c1)**2)

# Apply the Gaussian (RBF) function
Z = gaussian(r, sigma)

# Plot the 3D surface
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X0, X1, Z, cmap='viridis')

ax.set_title('Two-Input RBF Neuron Response')
ax.set_xlabel('Input x0')
ax.set_ylabel('Input x1')
ax.set_zlabel('Output y')

plt.show()
```
