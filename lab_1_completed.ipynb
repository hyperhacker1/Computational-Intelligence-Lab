# Task 1

Modify previous example and define function for testing if the number is a prime number.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

for i in range(1, 101):
    if is_prime(i):
        print(i)
```

# Task 2

Modify the previous task and define class finding the prime numbers in selected range.


```python
class PrimeNumbers:
    def __init__(self, start, end):
        self.start = start
        self.end = end

    def find_primes(self):
        primes = []
        for num in range(self.start, self.end + 1):
            if num < 2:
                continue
            prime = True
            for i in range(2, num):
                if num % i == 0:
                    prime = False
                    break
            if prime:
                primes.append(num)
        return primes

numbers = PrimeNumbers(1, 100)
print(numbers.find_primes())
```

# Task 3

Load Iris dataset from https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data.


```python
import numpy as np
import pandas as pd

iris_dataset = []
with open('iris.data', 'r') as file:
    for line in file:
        line = line.strip()
        if line != '':
            row = line.split(',')
            iris_dataset.append(row)

print('Loaded rows:', len(iris_dataset))
```

## Task 5

Rencode labels to three numerical colums according to the table:

Label | Output[0] | Output[1] | Output[2]
:--- | :---: | :---: | :---: |
Iris-versicolor | 1 | 0 | 0
Iris-virginica  | 0 | 1 | 0
Iris-setosa     | 0 | 0 | 1

so that the first row of the Iris dataset will look like ``5.1 3.5 1.4 0.2 0 0 1``  



```python
iris_dataset_enc = []
for row in iris_dataset:
    values = [float(row[0]), float(row[1]), float(row[2]), float(row[3])]
    label = row[4]

    if label == 'Iris-versicolor':
        values = values + [1, 0, 0]
    elif label == 'Iris-virginica':
        values = values + [0, 1, 0]
    else:
        values = values + [0, 0, 1]

    iris_dataset_enc.append(values)

print(iris_dataset_enc)
```

Split the ``iris_dataset_enc`` into two numpy arrays. The ``iris_features`` will contain four numerical columns containing features and the ``iris_desired`` will contain three numerical outputs containing desired outputs.

```python
iris_features = np.array([row[:4] for row in iris_dataset_enc], dtype=float)
iris_desired = np.array([row[4:] for row in iris_dataset_enc], dtype=int)

print(iris_features.shape)
print(iris_desired.shape)
```

```python
print('Features:')
print(iris_features)
print('Desired outputs:')
print(iris_desired)
```

## Task 7

Normalize features in the ``iris_dataset`` into the range of $<0,1>$ using min-max norm according to the following formula:

$$ y = \frac{y_{max} - y_{min}}{x_{max}-x_{min}}(x - x_{min}) + y_{min} $$

where $x$ is raw value and $y$ is the normalised value. The $y_{min}$ and $y_{max}$ determine the output range. The $x_{max}$ and $x_{min}$ denote the input range.

For this task $y_{max}=1$, $y_{min}=0$. The $x_{max}$ and $x_{min}$ must be obtained from the dataset, for each column separately.

The result prints with limited precision to two decimal digits (0.00).

```python
iris_values = np.array([[float(row[i]) for i in range(4)] for row in iris_dataset])
x_min = iris_values.min(axis=0)
x_max = iris_values.max(axis=0)

iris_normalized = (iris_values - x_min) / (x_max - x_min)
print(np.round(iris_normalized, 2))
```

```python
print('Minimum values:', x_min)
print('Maximum values:', x_max)
```

# Task 9

The sigmoid function is frequently used activation of a neuron. It is defined by

$$y=\sigma(x)=\frac{1}{1+e^{-x}}$$

Draw a graf of this function for $x\in <-8,8>$.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-8, 8, 200)
y = 1 / (1 + np.exp(-x))

plt.plot(x, y)
plt.title('Sigmoid function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

# Task 11

The Gaussian function is frequently used as an activation function of RBF neurons. It is defined by the equation

$$y=g(x)=e^{\frac{-x^2}{2\sigma^2}}$$

Draw a graf of this function for $x\in <-10,10>$ and $\sigma=\{0.2,0.5,1,2,5\}$.

```python
x = np.linspace(-10, 10, 300)
sigmas = [0.2, 0.5, 1, 2, 5]

for sigma in sigmas:
    y = np.exp(-(x ** 2) / (2 * sigma ** 2))
    plt.plot(x, y, label='sigma = ' + str(sigma))

plt.title('Gaussian functions')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
```

# Task 12

Draw response of the two input perceptron neuron defined by

$$y=\sigma(\sum_{i=0}^{1}w_ix_i + b)$$

where $w_0=10$,$w_1=10$, $b=-10$.

```python
x0 = np.linspace(0, 1, 100)
x1 = np.linspace(0, 1, 100)
X0, X1 = np.meshgrid(x0, x1)

w0 = 10
w1 = 10
b = -10
z = w0 * X0 + w1 * X1 + b
Y = 1 / (1 + np.exp(-z))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X0, X1, Y, cmap='viridis')
ax.set_xlabel('x0')
ax.set_ylabel('x1')
ax.set_zlabel('output')
plt.title('Two input perceptron')
plt.show()
```

# Task 13

Draw response of the two input RBF neuron defined by

$$y=g(\sqrt{\sum_{i=0}^{1}(x_i-c_i)^2})$$

where $c_0=0.5$, $c_1=0.5$, $\sigma=0.15$.

```python
x0 = np.linspace(0, 1, 100)
x1 = np.linspace(0, 1, 100)
X0, X1 = np.meshgrid(x0, x1)

c0 = 0.5
c1 = 0.5
sigma = 0.15
distance = np.sqrt((X0 - c0) ** 2 + (X1 - c1) ** 2)
Y = np.exp(-(distance ** 2) / (2 * sigma ** 2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X0, X1, Y, cmap='viridis')
ax.set_xlabel('x0')
ax.set_ylabel('x1')
ax.set_zlabel('output')
plt.title('RBF neuron')
plt.show()
```
